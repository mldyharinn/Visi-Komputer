```python
import numpy as np
from tensorflow.keras.datasets import mnist
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalisasi data
x_train = x_train / 255.0
x_test = x_test / 255.0

# Flatten gambar 28x28 menjadi 784 fitur
x_train_flat = x_train.reshape(x_train.shape[0], -1)
x_test_flat = x_test.reshape(x_test.shape[0], -1)

print("Ukuran training setelah flatten:", x_train_flat.shape)
print("Ukuran testing setelah flatten :", x_test_flat.shape)

# Menggunakan 5000 data training
x_train_svm = x_train_flat[:5000]
y_train_svm = y_train[:5000]

print("Ukuran data training untuk SVM:", x_train_svm.shape)

# Membuat model SVM dengan kernel linear
svm_model = SVC(kernel='linear')

# Training model
svm_model.fit(x_train_svm, y_train_svm)

# Prediksi data testing
y_pred = svm_model.predict(x_test_flat)

# Menghitung akurasi
accuracy = accuracy_score(y_test, y_pred)

print("Akurasi SVM:", accuracy)
```

```python
# ==========================================
# D3 - SIMPLE CNN
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Load dataset MNIST
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# 2. Normalisasi data
x_train = x_train / 255.0
x_test = x_test / 255.0

# 3. Menambahkan channel untuk CNN
x_train_cnn = x_train[..., np.newaxis]
x_test_cnn = x_test[..., np.newaxis]

print("Ukuran data training:", x_train_cnn.shape)
print("Ukuran data testing :", x_test_cnn.shape)

# 4. Membuat arsitektur CNN
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 5. Melihat struktur model
model.summary()

# 6. Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 7. Training model
history = model.fit(
    x_train_cnn,
    y_train,
    epochs=5,
    validation_split=0.1
)

# 8. Evaluasi model pada data testing
test_loss, test_accuracy = model.evaluate(
    x_test_cnn,
    y_test,
    verbose=0
)

print("Akurasi data testing:", test_accuracy)

# 9. Grafik akurasi
plt.figure(figsize=(8,5))
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Grafik Akurasi CNN')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# 10. Grafik loss
plt.figure(figsize=(8,5))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Grafik Loss CNN')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

from google.colab import files

uploaded = files.upload()
```

```python
# ==========================================
# D5 - TRANSFER LEARNING VGG16
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16

# 1. Load dataset CIFAR-10
(x_train_vgg, y_train_vgg), (x_test_vgg, y_test_vgg) = tf.keras.datasets.cifar10.load_data()

# Mengubah label menjadi 1 dimensi
y_train_vgg = y_train_vgg.flatten()
y_test_vgg = y_test_vgg.flatten()

# 2. Normalisasi data
x_train_vgg = x_train_vgg / 255.0
x_test_vgg = x_test_vgg / 255.0

print("Ukuran data training:", x_train_vgg.shape)
print("Ukuran data testing :", x_test_vgg.shape)

# 3. Membuat base model VGG16
base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(32, 32, 3)
)

# Membekukan base model
base_model.trainable = False

# 4. Membuat model transfer learning
model_vgg = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 5. Melihat struktur model
model_vgg.summary()

# 6. Compile model
model_vgg.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 7. Training model
history_vgg = model_vgg.fit(
    x_train_vgg,
    y_train_vgg,
    epochs=5,
    validation_split=0.1
)

# 8. Evaluasi model
test_loss_vgg, test_accuracy_vgg = model_vgg.evaluate(
    x_test_vgg,
    y_test_vgg,
    verbose=0
)

print("Akurasi testing VGG16:", test_accuracy_vgg)

# 9. Grafik Accuracy
plt.figure(figsize=(8,5))
plt.plot(
    history_vgg.history['accuracy'],
    label='Training Accuracy'
)
plt.plot(
    history_vgg.history['val_accuracy'],
    label='Validation Accuracy'
)
plt.title('Grafik Akurasi Transfer Learning VGG16')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# 10. Grafik Loss
plt.figure(figsize=(8,5))
plt.plot(
    history_vgg.history['loss'],
    label='Training Loss'
)
plt.plot(
    history_vgg.history['val_loss'],
    label='Validation Loss'
)
plt.title('Grafik Loss Transfer Learning VGG16')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16

# 1. Load dataset CIFAR-10
(x_train_vgg, y_train_vgg), (x_test_vgg, y_test_vgg) = tf.keras.datasets.cifar10.load_data()

# Mengubah label menjadi 1 dimensi
y_train_vgg = y_train_vgg.flatten()
y_test_vgg = y_test_vgg.flatten()

# 2. Normalisasi data
x_train_vgg = x_train_vgg / 255.0
x_test_vgg = x_test_vgg / 255.0

# 3. Membuat base model VGG16
base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(32, 32, 3)
)

# Membekukan base model
base_model.trainable = False

# 4. Membuat model transfer learning
model_vgg = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 5. Compile model (initial compile for the transfer learning model)
model_vgg.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ==========================================
# TUGAS KECIL D5 - FINE-TUNING VGG16
# ==========================================

# Membuka base model agar bisa dilakukan fine-tuning
base_model.trainable = True

# Membekukan semua layer kecuali 2 layer terakhir
for layer in base_model.layers[:-2]:
    layer.trainable = False

# Menampilkan status trainable setiap layer
for layer in base_model.layers:
    print(layer.name, ":", layer.trainable)

# Compile ulang model setelah mengubah trainable layer
model_vgg.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Training ulang dengan fine-tuning
history_vgg_finetune = model_vgg.fit(
    x_train_vgg,
    y_train_vgg,
    epochs=5,
    validation_split=0.1
)

# Evaluasi model
test_loss_finetune, test_accuracy_finetune = model_vgg.evaluate(
    x_test_vgg,
    y_test_vgg,
    verbose=0
)

print("Akurasi testing setelah fine-tuning:", test_accuracy_finetune)

# Grafik Accuracy
plt.figure(figsize=(8,5))
plt.plot(
    history_vgg_finetune.history['accuracy'],
    label='Training Accuracy'
)
plt.plot(
    history_vgg_finetune.history['val_accuracy'],
    label='Validation Accuracy'
)
plt.title('Grafik Akurasi VGG16 Setelah Fine-Tuning')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# Grafik Loss
plt.figure(figsize=(8,5))
plt.plot(
    history_vgg_finetune.history['loss'],
    label='Training Loss'
)
plt.plot(
    history_vgg_finetune.history['val_loss'],
    label='Validation Loss'
)
plt.title('Grafik Loss VGG16 Setelah Fine-Tuning')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
# ==========================================
# D6 - EVALUASI MODEL CNN CIFAR-10
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# ==========================================
# 1. Load dataset CIFAR-10
# ==========================================

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Ubah label menjadi 1 dimensi
y_train = y_train.flatten()
y_test = y_test.flatten()

# Normalisasi
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Gunakan sebagian data agar proses lebih cepat
x_train_d6 = x_train[:2000]
y_train_d6 = y_train[:2000]

x_test_d6 = x_test[:1000]
y_test_d6 = y_test[:1000]

print("Data training :", x_train_d6.shape)
print("Data testing  :", x_test_d6.shape)


# ==========================================
# 2. Membuat CNN
# ==========================================

model_d6 = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu',
                  input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])


# ==========================================
# 3. Compile model
# ==========================================

model_d6.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# ==========================================
# 4. Training
# ==========================================

history_d6 = model_d6.fit(
    x_train_d6,
    y_train_d6,
    epochs=5,
    validation_split=0.1,
    verbose=1
)


# ==========================================
# 5. Evaluasi akurasi
# ==========================================

test_loss, test_accuracy = model_d6.evaluate(
    x_test_d6,
    y_test_d6,
    verbose=0
)

print("\nAkurasi testing:", test_accuracy)


# ==========================================
# 6. Prediksi
# ==========================================

y_pred = model_d6.predict(
    x_test_d6,
    verbose=0
).argmax(axis=1)


# ==========================================
# 7. Classification Report
# ==========================================

class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]

print("\n==========================================")
print("CLASSIFICATION REPORT")
print("==========================================")

print(classification_report(
    y_test_d6,
    y_pred,
    target_names=class_names
))


# ==========================================
# 8. Confusion Matrix
# ==========================================

cm = confusion_matrix(y_test_d6, y_pred)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix CNN CIFAR-10")

plt.show()
```

```python
# ============================================================
# D3 - CNN MNIST
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist

# Load dataset MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalisasi 0-1
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Tambahkan channel
x_train = x_train[..., np.newaxis]
x_test = x_test[..., np.newaxis]

# Membuat model CNN
model = models.Sequential([
    layers.Conv2D(
        32,
        (3, 3),
        activation='relu',
        input_shape=(28, 28, 1)
    ),

    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(64, activation='relu'),

    layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Training
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1
)

# Evaluasi
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print(f"Akurasi testing: {test_accuracy:.4f}")
```

```python

# 1. Import library
from google.colab import files
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps


# ============================================================
# 2. Upload gambar tulisan tangan
# ============================================================

uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("File yang digunakan:", filename)


# ============================================================
# 3. Membuka dan melakukan preprocessing gambar
# ============================================================

# Membuka gambar
img = Image.open(filename)

# Mengubah gambar menjadi grayscale
img = img.convert('L')

# Meningkatkan kontras gambar
img = ImageOps.autocontrast(img)

# Mengubah gambar menjadi array
arr = np.array(img).astype(np.uint8)


# ============================================================
# 4. Membalik warna jika background gambar terang
# ============================================================

if arr.mean() > 127:
    img = ImageOps.invert(img)

arr = np.array(img)


# ============================================================
# 5. Mencari dan memotong bagian yang berisi tulisan
# ============================================================

thr = np.mean(arr) * 0.8
mask = arr > thr

if mask.any():
    ys, xs = np.where(mask)

    y0, y1 = ys.min(), ys.max()
    x0, x1 = xs.min(), xs.max()

    img = img.crop((x0, y0, x1 + 1, y1 + 1))


# ============================================================
# 6. Mengubah ukuran tulisan maksimal 20x20 piksel
# ============================================================

img.thumbnail((20, 20), Image.Resampling.LANCZOS)

w, h = img.size


# ============================================================
# 7. Membuat canvas berukuran 28x28 piksel
# ============================================================

canvas = Image.new('L', (28, 28), color=0)

# Menempatkan tulisan di bagian tengah canvas
canvas.paste(
    img,
    ((28 - w) // 2, (28 - h) // 2)
)


# ============================================================
# 8. Normalisasi nilai piksel menjadi 0-1
# ============================================================

arr = np.array(canvas).astype('float32') / 255.0


# ============================================================
# 9. Menampilkan hasil preprocessing
# ============================================================

plt.figure(figsize=(4, 4))

plt.imshow(arr, cmap='gray')

plt.title('Hasil Pre-processing 28x28')

plt.axis('off')

plt.show()


# Menampilkan ukuran gambar
print("Shape gambar:", arr.shape)


# ============================================================
# 10. Menyesuaikan bentuk gambar dengan input CNN
# ============================================================

# Menambahkan dimensi batch dan channel
img_cnn = arr.reshape(1, 28, 28, 1)

print("Shape input CNN:", img_cnn.shape)


# ============================================================
# 11. Melakukan prediksi menggunakan model CNN D3
# ============================================================

prediction = model.predict(img_cnn)

# Mengambil kelas dengan probabilitas tertinggi
predicted_digit = np.argmax(prediction)


# ============================================================
# 12. Menampilkan hasil prediksi
# ============================================================

print("\n========================================")
print("HASIL KLASIFIKASI")
print("========================================")

print("Hasil prediksi:", predicted_digit)


# ============================================================
# 13. Menampilkan probabilitas setiap angka
# ============================================================

print("\nProbabilitas setiap angka:")

for i, probability in enumerate(prediction[0]):
    print(f"Angka {i}: {probability:.4f}")
```
